### Задача 1: по теореме сложения вероятностей



```python
# Данные задачи
total_balls = 10  # Общее количество шаров
red_balls = 5     # Количество красных шаров
green_balls = 2   # Количество зелёных шаров

# Вероятности событий A и B
P_A = red_balls / total_balls  # Вероятность достать красный шар
P_B = green_balls / total_balls  # Вероятность достать зелёный шар

# Поскольку красный и зелёный шар — взаимоисключающие события, P(A ∩ B) = 0
P_A_and_B = 0

# Теорема сложения вероятностей
P_A_or_B = P_A + P_B - P_A_and_B

# Вывод результатов
print("Вероятность достать красный шар (P(A)):", round(P_A, 2))
print("Вероятность достать зелёный шар (P(B)):", round(P_B, 2))
print("Вероятность достать красный или зелёный шар (P(A ∪ B)):", round(P_A_or_B, 2))
```

### Задача 2: по теореме умножения вероятностей 

### для независимых событий



```python
# Данные задачи
red_box1 = 3     # Количество красных шаров в первой коробке
total_box1 = 10  # Общее количество шаров в первой коробке
red_box2 = 4     # Количество красных шаров во второй коробке
total_box2 = 10  # Общее количество шаров во второй коробке

# Вероятности
P_A = red_box1 / total_box1  # Вероятность достать красный шар из первой коробки
P_B = red_box2 / total_box2  # Вероятность достать красный шар из второй коробки

# Используем теорему умножения для независимых событий
P_A_and_B = P_A * P_B

# Вывод результата
print("Вероятность достать красный шар из первой коробки (P(A)):", round(P_A, 2))
print("Вероятность достать красный шар из второй коробки (P(B)):", round(P_B, 2))
print("Вероятность достать красные шары из обеих коробок (P(A ∩ B)):", round(P_A_and_B, 4))
```

### Задача 3: по теореме умножения вероятностей 

### для зависимых событий



```python
# Данные задачи
red_balls = 3     # Количество красных шаров
total_balls = 10  # Общее количество шаров

# Вероятности
P_A = red_balls / total_balls  # Вероятность достать первый красный шар
P_B_given_A = (red_balls - 1) / (total_balls - 1)  # Вероятность достать второй красный шар после первого

# Используем теорему умножения для зависимых событий
P_A_and_B = P_A * P_B_given_A

# Вывод результата
print("Вероятность достать первый красный шар (P(A)):", round(P_A, 2))
print("Вероятность достать второй красный шар после первого (P(B|A)):", round(P_B_given_A, 2))
print("Вероятность достать два красных шара (P(A ∩ B)):", round(P_A_and_B, 4))
```

### Задача 1: Вычислить математическое ожидание


```python
# Возможные выигрыши
outcomes = [0, 5, 10]  # Значения x_i

# Соответствующие вероятности
probabilities = [2/6, 2/6, 2/6]  # P(x_i)

# Вычисление математического ожидания
expected_value = sum(x * p for x, p in zip(outcomes, probabilities))

# Вывод результата
print("Математическое ожидание выигрыша (E[X]):", round(expected_value, 2))
```

### Задача 2: Вычислить дисперсию


```python
# Возможные выигрыши
outcomes = [0, 5, 10]  # Значения x_i

# Соответствующие вероятности
probabilities = [2/6, 2/6, 2/6]  # P(x_i)

# Вычисление математического ожидания E[X]
expected_value = sum(x * p for x, p in zip(outcomes, probabilities))

# Вычисление E[X^2]
expected_value_squared = sum((x**2) * p for x, p in zip(outcomes, probabilities))

# Вычисление дисперсии Var(X)
variance = expected_value_squared - expected_value**2

# Вывод результатов
print("Математическое ожидание (E[X]):", round(expected_value, 2))
print("Математическое ожидание квадрата (E[X^2]):", round(expected_value_squared, 2))
print("Дисперсия (Var(X)):", round(variance, 2))
```

### Задача 3: Биноминальное распределение


```python
import matplotlib.pyplot as plt
import math

# Функция для вычисления биномиального распределения
def binomial_probability(n, k, p):
    binom_coeff = math.comb(n, k)  # Вычисление биномиального коэффициента
    return binom_coeff * (p**k) * ((1-p)**(n-k))

# Параметры задачи
n = 10  # Общее количество испытаний (пациенты)
p = 0.7  # Вероятность успеха (выздоровления)

# 1. Вычисление вероятности P(X = 7)
k = 7  # Количество успехов
probability_k_7 = binomial_probability(n, k, p)
print(f"Вероятность того, что ровно {k} пациентов выздоровеют: {probability_k_7:.4f}")

# 2. Построение графика распределения
probabilities = [binomial_probability(n, k, p) for k in range(n + 1)]

# График
plt.bar(range(n + 1), probabilities, color='skyblue', edgecolor='black')
plt.title("Биномиальное распределение (n=10, p=0.7)")
plt.xlabel("Количество успехов (k)")
plt.ylabel("Вероятность (P(X=k))")
plt.xticks(range(n + 1))
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### Задача 4: Нормальное распределение


```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# Параметры нормального распределения
mu = 70      # Среднее значение
sigma = 10   # Стандартное отклонение

# 1. Вероятность, что оценки в диапазоне от 60 до 80
probability = stats.norm.cdf(80, mu, sigma) - stats.norm.cdf(60, mu, sigma)
print(f"Вероятность, что оценка в диапазоне от 60 до 80: {probability:.4f}")

# 2. Построение графика плотности вероятности
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)  # Диапазон значений
pdf = stats.norm.pdf(x, mu, sigma)  # Плотность вероятности

# Построение графика
plt.plot(x, pdf, label='Нормальное распределение', color='blue')
plt.fill_between(x, pdf, where=(x >= 60) & (x <= 80), color='skyblue', alpha=0.5, label='Диапазон 60-80')
plt.title('График плотности нормального распределения')
plt.xlabel('Оценки')
plt.ylabel('Плотность вероятности')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

# Lesson topic: Теорема Байеса

В основе анализа данных, машинного обучения и работы над искусственным интеллектом лежат математика и статистика. 

Аналитики данных и специалисты по Data Science часто обращаются к теории вероятностей, чтобы принимать решения в условиях неопределённости. 

Например, выбрать наиболее удачную гипотезу на старте проекта, когда данных ещё нет. 

Один из прикладных инструментов теории вероятностей — теорема Байеса. 

Её назвали в честь автора — английского священника и математика Томаса Байеса. 

В работе, опубликованной в 1763 году, Байес предложил теорему для корректировки убеждений при появлении новых данных.

В основе теоремы — условная вероятность. То есть при наличии двух связанных между собой событий с помощью теоремы можно рассчитать вероятность 

наступления одного из них при условии, что наступило другое. 

При расчёте вероятности с помощью теоремы Байеса можно поменять причину и следствие местами. 

Например, кашель — это следствие заболевания. Допустим, исследователь собрал данные о том, как часто у пациентов с бронхитом есть кашель. 

С помощью теоремы можно рассчитать вероятность бронхита при наличии у пациента кашля.

### Формула Байеса

Вот как выглядит формула теоремы Байеса:



Расшифруем все обозначения:

##### ●	P — вероятность.

##### ●	| — математическое обозначение фразы «при условии, что». 

##### ●	P(B|A) — вероятность наступления события B при условии, что произошло выполняется гипотеза A.

##### ●	P(A|B) — апостериорная вероятность. 

##### То есть вероятность события с учётом апостериорных данных — данных, которые получены после наблюдения или опыта.

##### ●	P(A) — априорная, или безусловная, вероятность наступления события А. Это вероятность, 

##### которая основана на имеющихся данных до наблюдения или проведения опыта.

##### ●	P(B) — априорная вероятность наступления события В.

#### Априорная — безусловная — вероятность пересматривается при поступлении новых данных и становится апостериорной — условной. 

На практике это выглядит примерно так: сначала есть убеждение или гипотеза, затем в ходе наблюдения, 

опыта или эксперимента собирают новые данные, а после на их основе обновляют первоначальное убеждение.

Рассмотрим, как это происходит, на известной задаче теории вероятностей — парадоксе Монти Холла.

#### Парадокс Монти Холла: пример визуализации вероятности

Парадокс Монти Холла — задача с неожиданным, на первый взгляд, решением. 

Названа в честь американского ведущего развлекательного шоу «Давайте заключим сделку», 

поскольку условие задачи описано как ситуация из этого самого шоу. Есть несколько версий задачи. 

Рассмотрим самую популярную — с дополнительными условиями.

Условие задачи:

Перед участником игры три двери. 

За одной из них находится автомобиль, за двумя другими — козы. 

Участнику нужно угадать, за какой дверью автомобиль. 

Он выбирает дверь под номером три. Прежде чем открыть третью дверь, ведущий открывает вторую. 

За ней оказывается коза. 



После этого ведущий спрашивает участника, не хочет ли тот изменить своё первоначальное решение. 

При этом участнику игры известно, что:

●	ведущий знает, за какой дверью находится автомобиль;

●	ведущий должен открыть дверь, за которой находится коза, но не ту, которую выбрал участник, и предложить изменить выбор;

●	если участник указал на дверь, за которой находится автомобиль, ведущий выбирает любую из оставшихся с одинаковой вероятностью.

Возрастут ли шансы участника отгадать дверь с автомобилем, если он изменит своё первоначальное решение?

Решим задачу с помощью теоремы Байеса: 

##### Шаг 1. Введём обозначения

Пусть: 

●	A1, A2, A3 — события, при которых автомобиль находится за дверью номер один, два и три соответственно. 

●	B1, B2, B3 — события, при которых ведущий открыл дверь номер один, два и три соответственно.

##### Шаг 2. Подставим обозначения в формулу теоремы Байеса
 
Чтобы оценить, повысятся ли шансы отгадать нужную дверь при смене первоначального решения, 

нужно рассчитать вероятность выигрыша при смене решения с учётом новых данных. 

То есть вероятность того, что автомобиль находится за первой дверью при условии, что ведущий открыл вторую дверь, — P(A1|B2). 

Формула Байеса для расчёта вероятности будет выглядеть так:



Чтобы рассчитать P(A1|B2), нужно вычислить P(A1), P(B2|A1) и P(B2). 

##### Шаг 3. Рассчитаем P(A1)

Автомобиль равновероятно находится за одной из трёх дверей, поэтому:

P(A1) = P(A2) = P(A3) = 1/3

##### Шаг 4. Рассчитаем P(B2|A1)

Вероятность того, что ведущий откроет вторую дверь при условии, что автомобиль за первой дверью, 

а участник выбрал вторую дверь, равна единице. Потому что у ведущего не остаётся выбора, какую дверь открыть.

P(B2|A1) = 1

##### Шаг 5. Рассчитаем P(B2)

P(B2) — это априорная вероятность того, что ведущий откроет вторую дверь. 

То есть вероятность с учётом всех возможных вариантов. Она рассчитывается так:

P(B2) = P(B2|A1) ⋅ P(A1) + P(B2|A2) ⋅ P(A2) + P(B2|A3) ⋅ P(A3)

Вероятность, что ведущий откроет дверь, за которой находится автомобиль, по условию задачи равна нулю. Поэтому:

P(B2|A2) ⋅ P(A2) = 0

Разберём, как рассчитываются остальные слагаемые при тех условиях, которые есть.

Значения P(B2|A1) и P(A1) уже известны из расчёта выше, поэтому:

P(B2|A2) ⋅ P(A2) = 0

Разберём, как рассчитываются остальные слагаемые при тех условиях, которые есть.

Значения P(B2|A1) и P(A1) уже известны из расчёта выше, поэтому:

P(B2|A1)⋅ P(A1) = 1⋅ 1/3 = 1/3

Если автомобиль находится за дверью под номером три, которую выбрал участник, тогда по условию задачи ведущий 

с одинаковой вероятностью может выбрать любую из оставшихся дверей. 

Это значит, что P(B2|A3) = 1/2. Тогда второе ненулевое слагаемое рассчитывается так:

P(B2|A3) • P(A3) = 1/2 • 1/3 = 1/6

Таким образом:

P(B2) = 1/3 + 1/6

Шаг 6. Подставим значения в формулу теоремы Байеса



Это значит, что при изменении первоначального решения вероятность угадать автомобиль повышается с 1/2 до 2/3.

#### В чём парадокс задачи

Когда ведущий открывает одну из дверей, на первый взгляд кажется, что шансы выиграть автомобиль сами собой повышаются с 1/3 до 1/2, 

а смена двери на них никак не повлияет. Но математические расчёты показывают, что при смене двери вероятность выиграть автомобиль повышается до  2/3. 

Причина в том, что учитываются новые данные — дверь, которую открыл ведущий и за которой точно нет автомобиля.

##### Суть теоремы Байеса в том, что новые данные не формируют убеждение с нуля, а уточняют те, которые уже имеются.

#### Практический пример применения теоремы Байеса в Data Science

На теореме Байеса основан наивный байесовский классификатор — одна из основных моделей машинного обучения. 

Однако между теоремой Байеса и наивным байесовским классификатором есть важное отличие: условная независимость. 

Если в теореме речь идёт о взаимосвязанных событиях, в наивном классификаторе, наоборот, 

все объекты — события или свойства — независимы. 

Это может не соответствовать реальности, но такой «наивный» подход упрощает расчёт вероятностей и ускоряет обучение модели.

### Задача 1: Найти апостериорную вероятность


```python
# Исходные данные
P_C1 = 0.1  # Вероятность быть больным
P_C2 = 0.9  # Вероятность быть здоровым
P_positive_given_C1 = 0.95  # Вероятность положительного теста при болезни
P_positive_given_C2 = 0.05  # Вероятность положительного теста при здоровье

# Общая вероятность положительного теста (нормализация)
P_positive = P_positive_given_C1 * P_C1 + P_positive_given_C2 * P_C2

# Апостериорная вероятность (по формуле Байеса)
P_C1_given_positive = (P_positive_given_C1 * P_C1) / P_positive

# Вывод результатов
print("Вероятность быть больным при положительном тесте (P(C1|+)):")
print(f"{P_C1_given_positive:.4f} или {P_C1_given_positive * 100:.2f}%")
```

### Задача 2: Теорема Байеса в машинном обучении (Email Spam Detection)



```python
# Данные задачи
P_S = 0.3                  # Вероятность, что письмо является спамом (P(S))
P_not_S = 1 - P_S          # Вероятность, что письмо не является спамом (P(¬S))
P_W_given_S = 0.8          # Вероятность слова "дешево" в спаме (P(W | S))
P_W_given_not_S = 0.1      # Вероятность слова "дешево" в не-спаме (P(W | ¬S))

# Полная вероятность того, что слово "дешево" встречается в письме (P(W))
P_W = P_W_given_S * P_S + P_W_given_not_S * P_not_S

# Вероятность, что письмо — спам, если оно содержит слово "дешево" (P(S | W))
P_S_given_W = (P_W_given_S * P_S) / P_W

# Вывод результата
print(f"Вероятность, что письмо — спам, если оно содержит слово 'дешево': {P_S_given_W:.4f}")
```
