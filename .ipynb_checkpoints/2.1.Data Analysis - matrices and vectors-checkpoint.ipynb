На Python действия с векторами и матрицами реализованы с помощью библиотеки Numpy.

### Задача 1: создадим матрицу 2 Х 2 и посчитаем её определитель

[[2,1],[2,3]]

##### np.linalg — это модуль для операций из линейной алгебры в библиотеке NumPy. 

Этот пакет позволяет работать с векторами, матрицами и другими линейными структурами данных. 

Некоторые функции из пакета np.linalg:

##### ● np.linalg.det() — нахождение определителя матрицы.

##### ● np.linalg.inv() — нахождение обратной матрицы. 

##### ● np.linalg.solve() — решение систем линейных уравнений. 

##### ● np.linalg.eig() — нахождение собственных значений и собственных векторов.

```python
# импортируем библиотеку numpy
import numpy as np
import pandas as pd

a = np.array([[2,1],[2,3]])
print(a)
```

```python
# найдём определитель матрицы
np.linalg.det(a)
```

### Задача 2: найдём обратную матрицу

# Задача 1: Решим систему уравнений:

### 2x+3y=8
### x−y=1

```python
import numpy as np

# Коэффициенты системы уравнений
A = np.array([[2, 3],  # Коэффициенты уравнения 1
              [1, -1]])  # Коэффициенты уравнения 2

# Вектор свободных членов
B = np.array([8, 1])

# Решение системы
solution = np.linalg.solve(A, B)
print("Решение системы:")
print(solution)
```

### Задача 2: решение системы Неоднородных Линейных Уравнений

x = np.array([[2,1],[2,3]])

y = np.array([5,-10])

Здесь:

X - матрица,

w - вектор,

y - вектор.

В нашем случае мы уже имеем матрицу a, которая при решении системы НЛУ, это матрица X.

```python
x = np.array([[2,1],[2,3]])
b = np.linalg.inv(x)
```

```python
# Пусть нам дан вектор y
y = np.array([5,-10])
```

```python
# Найдём решение системы НЛУ
w = np.linalg.solve(x,y)
print(w)
```

```python
# сделаем проверку для найденного решения
print(b.dot(y))
```

### Задача 3: решение системы Неоднородных Линейных Уравнений

### Пример системы:

### 3x+2y−z=1

### 2x−2y+4z=−2

### −x+ 0.5​y−z=0

```python
import numpy as np

# Матрица коэффициентов системы
A = np.array([
    [3, 2, -1],   # Коэффициенты уравнения 1
    [2, -2, 4],   # Коэффициенты уравнения 2
    [-1, 0.5, -1] # Коэффициенты уравнения 3
])

# Вектор свободных членов
B = np.array([1, -2, 0])

# Решение системы
solution = np.linalg.solve(A, B)
print("Решение системы:")
print(solution)
```

### Задача 2: Создадим два вектора и найдём векторное произведение 2-х векторов

[1, 2, 3]

[4, 5, 6]

```python
import numpy as np

# Создадим два 3D-вектора
vector_a = np.array([1, 2, 3])  # Вектор a
vector_b = np.array([4, 5, 6])  # Вектор b

# Найдём векторное произведение
cross_product = np.cross(vector_a, vector_b)

# Вывод результата
print("Вектор a:", vector_a)
print("Вектор b:", vector_b)
print("Векторное произведение:", cross_product)
```

### Задача 3: найти угол между векторами

[1.72, 54, 36.2]

[1.56, 47, 30.0]

Вспомним, чему равен косинус угла между двумя векторами.



```python
# даны два вектора (данные двух человек)
# нужно понять, насколько они схожи
b = np.array([1.72, 54, 36.2])
c = np.array([1.56, 47, 30.0])
 
# вначале выполним операции в числителе формулы
numerator = np.dot(b, c)
 
# теперь займемся знаменателем и
# (1) рассчитаем длины (по большому счету, это теорема Пифагора)
bLen = np.linalg.norm(b)
cLen = np.linalg.norm(c)
 
# (2) перемножим их
denominator = bLen * cLen
 
# посмотрим, чему равен косинус угла между векторами
cosine = numerator/denominator
cosine
```

```python
# Расчитаем угол в градусах между двумя этими векторами
# для этого вначале вычислим угол в радианах
angle_radians = np.arccos(cosine)
 
# затем в градусах
angle_degrees = angle_radians * 360/2/np.pi
angle_degrees
```

Как мы видим, угол чрезвычайно мал, а значит векторы очень близки друг к другу.

Мы будем на практике применять косинусное сходство или близость векторов, например, 

при построении рекомендательных систем для сопутствующих товаров в интернет-магазине. 

Когда вы покупаете, например, холодильник, то микроволновую печь вам рекомендуют именно потому, 

что вектор ваших покупок очень близок к вектору покупок тех, кто купил микроволновку.
